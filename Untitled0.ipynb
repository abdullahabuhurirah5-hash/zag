```python

# Python typing
from typing import Any, Dict, Optional, Literal

# SQLAlchemy core
from sqlalchemy import Column, ForeignKey, Integer, String, Boolean, inspect, MetaData

# SQLAlchemy ORM
from sqlalchemy.ext.declarative import as_declarative, declared_attr
from sqlalchemy.orm import relationship

# Utility
import humps

# Shared metadata
metadata = MetaData()


@as_declarative(metadata=metadata)
class Base_Model:
    __name__: str

    @declared_attr
    def __tablename__(cls) -> str:
        return humps.depascalize(cls.__name__)

    def to_dict(self) -> Dict[str, Any]:
        return {
            attr.key: getattr(self, attr.key)
            for attr in inspect(self).mapper.column_attrs
        }


# ----------------------------
# SQLAlchemy Models
# ----------------------------
class User(Base_Model):
    id = Column(Integer, primary_key=True, index=True)
    full_name = Column(String, index=True)
    email = Column(String, unique=True, nullable=False, index=True)
    hashed_password = Column(String, nullable=False)
    is_active = Column(Boolean, default=True)
    is_superuser = Column(Boolean, default=False)
    items = relationship("Item", back_populates="owner", lazy="selectin")


class Item(Base_Model):
    id = Column(Integer, primary_key=True, index=True)
    title = Column(String, index=True)
    description = Column(String, index=True)
    owner_id = Column(Integer, ForeignKey("user.id"))
    owner = relationship("User", back_populates="items")


# ----------------------------
# Pydantic Models
# ----------------------------
from pydantic import (
    BaseModel as PydanticBaseModel,
    EmailStr,
    ValidationError,
    ConfigDict,
)

class JobSchema(PydanticBaseModel):
    job_id: str


class AuthToken(PydanticBaseModel):
    access_token: str
    token_type: Literal["bearer"]


class AuthTokenPayload(PydanticBaseModel):
    user_id: Optional[int] = None


class UserCore(PydanticBaseModel):
    email: Optional[EmailStr] = None
    full_name: Optional[str] = None
    is_active: bool = True
    is_superuser: bool = False


class UserCreateSchema(UserCore):
    email: EmailStr
    password: str


class UserDBSchema(UserCore):
    hashed_password: str


class UserResponse(UserCore):
    id: int
    model_config = ConfigDict(from_attributes=True)


class UserUpdateSchema(UserCore):
    password: Optional[str] = None


class UserUpdateDBSchema(UserCore):
    hashed_password: str


# ----------------------------
# Application Settings
# ----------------------------
from pydantic_settings import BaseSettings
from pydantic import SecretStr, field_validator
from pydantic_core.core_schema import ValidationInfo


class Settings(BaseSettings):
    PROJECT_NAME: str

    POSTGRES_DB: str
    POSTGRES_HOST: str
    POSTGRES_USER: str
    POSTGRES_PASSWORD: SecretStr
    POSTGRES_URI: Optional[str] = None

    @field_validator("POSTGRES_URI", mode="before")
    @classmethod
    def build_postgres_uri(
        cls, v: Optional[str], info: ValidationInfo
    ) -> str:
        if isinstance(v, str):
            return v

        data = info.data

        password: SecretStr = data.get("POSTGRES_PASSWORD", SecretStr(""))
        return (
            f"postgresql+asyncpg://"
            f"{data.get('POSTGRES_USER')}:"
            f"{password.get_secret_value()}@"
            f"{data.get('POSTGRES_HOST')}/"
            f"{data.get('POSTGRES_DB')}"
        )

    FIRST_USER_EMAIL: EmailStr
    FIRST_USER_PASSWORD: SecretStr

    SECRET_KEY: SecretStr
    ACCESS_TOKEN_EXPIRE_MINUTES: int

    REDIS_HOST: str
    REDIS_PORT: int


# ----------------------------
# Settings Instance
# ----------------------------
settings = Settings(
    PROJECT_NAME="TestProject",
    POSTGRES_DB="test_db",
    POSTGRES_HOST="localhost",
    POSTGRES_USER="test_user",
    POSTGRES_PASSWORD="test_pass",
    FIRST_USER_EMAIL="test@example.com",
    FIRST_USER_PASSWORD="test_pass",
    SECRET_KEY="supersecret",
    ACCESS_TOKEN_EXPIRE_MINUTES=30,
    REDIS_HOST="localhost",
    REDIS_PORT=6379,
)


# ----------------------------
# Database Engine & Session
# ----------------------------
from sqlalchemy.ext.asyncio import AsyncSession, create_async_engine
from sqlalchemy.orm import sessionmaker

engine = create_async_engine(settings.POSTGRES_URI)
SessionLocal = sessionmaker(
    engine, expire_on_commit=False, class_=AsyncSession
)


# ----------------------------
# Security / JWT
# ----------------------------
from datetime import datetime, timedelta
from passlib.context import CryptContext
from jose import jwt

password_manager = CryptContext(schemes=["bcrypt"], deprecated="auto")

JWT_ALGO = "HS256"


def hash_password(plain: str) -> str:
    return password_manager.hash(plain)


def verify_password(plain: str, hashed: str) -> bool:
    return password_manager.verify(plain, hashed)


def generate_token(user: User, minutes: Optional[int] = None) -> str:
    expiry_minutes = minutes or settings.ACCESS_TOKEN_EXPIRE_MINUTES
    payload = {
        "user_id": str(user.id),
        "exp": datetime.utcnow() + timedelta(minutes=expiry_minutes),
    }
    return jwt.encode(
        payload,
        key=settings.SECRET_KEY.get_secret_value(),
        algorithm=JWT_ALGO,
    )


# ----------------------------
# Authentication
# ----------------------------
async def verify_user(
    session: AsyncSession, email: EmailStr, password: str
) -> Optional[User]:
    candidate = await crud_user.get(session, email=email)
    if candidate and verify_password(password, candidate.hashed_password):
        return candidate
    return None

from typing import Any, Dict, Generic, List, Optional, Type, TypeVar, Union

from pydantic import BaseModel as PydanticBaseModel, EmailStr, ValidationError
from sqlalchemy import select
from sqlalchemy.ext.asyncio import AsyncSession

# ----------------------------
# Type variables for generics
# ----------------------------
TModel = TypeVar("TModel")
TCreate = TypeVar("TCreate", bound=PydanticBaseModel)
TUpdate = TypeVar("TUpdate", bound=PydanticBaseModel)


# ----------------------------
# Generic CRUD Base
# ----------------------------
class AsyncCRUD(Generic[TModel, TCreate, TUpdate]):
    def __init__(self, model: Type[TModel]) -> None:
        self._model_class = model

    async def create(self, session: AsyncSession, obj_in: TCreate) -> TModel:
        data = obj_in.dict(exclude_unset=True)  # only use fields provided
        instance = self._model_class(**data)
        session.add(instance)
        await session.commit()
        await session.refresh(instance)  # refresh to get auto-generated fields like id
        return instance

    async def get(
        self, session: AsyncSession, *filters, **filter_by
    ) -> Optional[TModel]:
        query = select(self._model_class).filter(*filters).filter_by(**filter_by)
        result = await session.execute(query)
        return result.scalars().first()

    async def get_all(
        self, session: AsyncSession, *filters, offset: int = 0, limit: int = 100, **filter_by
    ) -> List[TModel]:
        query = (
            select(self._model_class)
            .filter(*filters)
            .filter_by(**filter_by)
            .offset(offset)
            .limit(limit)
        )
        result = await session.execute(query)
        return result.scalars().all()

    async def update(
        self,
        session: AsyncSession,
        *,
        db_obj: Optional[TModel] = None,
        obj_in: Union[TUpdate, Dict[str, Any]],
        **filter_by
    ) -> Optional[TModel]:
        db_obj = db_obj or await self.get(session, **filter_by)
        if db_obj:
            current_data = db_obj.to_dict()
            if isinstance(obj_in, dict):
                update_data = obj_in
            else:
                update_data = obj_in.dict(exclude_unset=True)
            for field, value in update_data.items():
                if field in current_data:
                    setattr(db_obj, field, value)
            session.add(db_obj)
            await session.commit()
        return db_obj

    async def delete(
        self, session: AsyncSession, *filters, db_obj: Optional[TModel] = None, **filter_by
    ) -> Optional[TModel]:
        db_obj = db_obj or await self.get(session, *filters, **filter_by)
        if db_obj:
            await session.delete(db_obj)
            await session.commit()
        return db_obj


# ----------------------------
# User-specific CRUD instance
# ----------------------------
UserCRUD = AsyncCRUD[User, UserDBSchema, UserUpdateDBSchema]
crud_user = UserCRUD(User)
```

```python
from typing import AsyncGenerator
from fastapi import Depends, HTTPException
from fastapi.security import OAuth2PasswordBearer
from jose import jwt, JWTError
from pydantic import ValidationError
from sqlalchemy.ext.asyncio import AsyncSession

# Import our existing objects
from __main__ import settings, SessionLocal, crud_user, User, JWT_ALGO
from __main__ import AuthTokenPayload as TokenData

# ----------------------------
# OAuth2 Scheme
# ----------------------------
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="/api/v1/login")


# ----------------------------
# Database Session Dependency
# ----------------------------
async def provide_session() -> AsyncGenerator[AsyncSession, None]:
    """Yields a database session for dependency injection."""
    async with SessionLocal() as session:
        yield session


# ----------------------------
# Token Validation Dependency
# ----------------------------
def extract_token_data(token: str = Depends(oauth2_scheme)) -> TokenData:
    """
    Decode JWT token and return payload as TokenData.
    Raises HTTPException(403) if token is invalid.
    """
    try:
        payload = jwt.decode(
            token,
            key=settings.SECRET_KEY.get_secret_value(),
            algorithms=[JWT_ALGO],
        )
        return TokenData(**payload)
    except (JWTError, ValidationError):
        raise HTTPException(status_code=403, detail="Invalid or expired token")


# ----------------------------
# Current User Dependency
# ----------------------------
async def fetch_current_user(
    token_data: TokenData = Depends(extract_token_data),
    session: AsyncSession = Depends(provide_session),
) -> User:
    """
    Retrieve the current user from the database.
    Raises HTTPException if user not found or inactive.
    """
    user = await crud_user.get(session, id=int(token_data.user_id))
    if not user:
        raise HTTPException(status_code=404, detail="User not found")
    if not user.is_active:
        raise HTTPException(status_code=400, detail="Inactive user")
    return user


# ----------------------------
# Superuser Check Dependency
# ----------------------------
def require_superuser(
    current_user: User = Depends(fetch_current_user),
) -> User:
    """Ensure the current user is a superuser."""
    if not current_user.is_superuser:
        raise HTTPException(
            status_code=403,
            detail="User does not have superuser privileges"
        )
    return current_user


# Alias for clarity
on_superuser = require_superuser
```

```python

# ----------------------------
# Health, Home, Auth, and Task Routers
# ----------------------------

import asyncio
import socket
from typing import Any

from fastapi import APIRouter, Depends, HTTPException
from fastapi.security import OAuth2PasswordRequestForm
from starlette.responses import Response
from sqlalchemy.ext.asyncio import AsyncSession

from arq.jobs import Job as ArqJob
import redis.asyncio as redis




redis_pool: redis.Redis | None = None

async def init_redis() -> redis.Redis:
    global redis_pool
    if redis_pool is None:
        # Use redis.asyncio directly
        redis_pool = redis.Redis(
            host=settings.REDIS_HOST,
            port=settings.REDIS_PORT,
            decode_responses=True
        )
    return redis_pool

# ----------------------------
# Health Router
# ----------------------------
health_router = APIRouter(prefix="/health", tags=["Health"])

@health_router.get("/", status_code=204)
async def check_health(session: AsyncSession = Depends(provide_session)) -> Response:
    """Check database connectivity and return status."""
    try:
        await asyncio.wait_for(session.execute("SELECT 1"), timeout=1)
    except (asyncio.TimeoutError, socket.gaierror):
        return Response(status_code=503)
    return Response(status_code=204)

# ----------------------------
# Home Router
# ----------------------------
home_router = APIRouter(prefix="/home", tags=["Home"])

@home_router.get("/", dependencies=[Depends(extract_token_data)])
async def greet_user() -> str:
    return "Hello World!"

@home_router.get("/another/")
async def greet_again() -> str:
    return "Another Hello World!"

# ----------------------------
# Authentication Router
# ----------------------------
auth_router = APIRouter(tags=["Login"])

@auth_router.post("/login/", response_model=dict)
async def login_user(
    credentials: OAuth2PasswordRequestForm = Depends(),
    session: AsyncSession = Depends(provide_session),
) -> dict:
    """
    Authenticate user and return access token.
    """
    user: User | None = await verify_user(session, credentials.username, credentials.password)
    if not user:
        raise HTTPException(status_code=400, detail="Incorrect email or password")
    if not user.is_active:
        raise HTTPException(status_code=400, detail="Inactive user")
    token = generate_token(user)
    return {"access_token": token, "token_type": "bearer"}

# ----------------------------
# Task / Background Job Router
# ----------------------------
tasks_router = APIRouter(prefix="/tasks", tags=["Tasks"])

@tasks_router.on_event("startup")
async def setup_redis() -> None:
    """Initialize Redis pool at startup."""
    await init_redis()

@tasks_router.post("/", response_model=dict, status_code=201)
async def enqueue_task(message: str) -> dict:
    """
    Create a background task with Redis/ARQ.
    """
    redis_conn = await init_redis()
    job_id = await redis_conn.execute_command("LPUSH", "task_queue", message)
    return {"id": str(job_id)}

@tasks_router.get("/{task_id}/")
async def fetch_task_info(task_id: str) -> Any:
    """
    Retrieve info about a specific task.
    """
    redis_conn = await init_redis()
    job_data = await redis_conn.lindex("task_queue", int(task_id))
    if job_data is None:
        raise HTTPException(status_code=404, detail="Task not found")
    return {"task_id": task_id, "message": job_data}
```

```python
from typing import List

from fastapi import APIRouter, Depends, HTTPException
from sqlalchemy.exc import IntegrityError
from sqlalchemy.ext.asyncio import AsyncSession

# Import our existing code
from __main__ import (
    User,
    UserCreateSchema,
    UserUpdateDBSchema,
    UserUpdateSchema,
    UserResponse,
    crud_user,
    provide_session,
    hash_password,
    fetch_current_user,
    on_superuser,
)

# ----------------------------
# Users Router
# ----------------------------
users_router = APIRouter(prefix="/users", tags=["Users"])


# ----------------------------
# Read all users (superuser only)
# ----------------------------
@users_router.get("/", response_model=List[UserResponse], dependencies=[Depends(on_superuser)])
async def read_users(offset: int = 0, limit: int = 100, session: AsyncSession = Depends(provide_session)):
    """
    Retrieve users.
    """
    users = await crud_user.get_all(session, offset=offset, limit=limit)
    return users


# ----------------------------
# Create a new user (superuser only)
# ----------------------------
@users_router.post("/", response_model=UserResponse, dependencies=[Depends(on_superuser)])
async def create_user(user_in: UserCreateSchema, session: AsyncSession = Depends(provide_session)):
    """
    Create new user.
    """
    existing_user = await crud_user.get(session, email=user_in.email)
    if existing_user:
        raise HTTPException(
            status_code=409,
            detail="The user with this email already exists in the system"
        )
    db_obj = UserUpdateDBSchema(**user_in.dict(), hashed_password=hash_password(user_in.password))
    new_user = await crud_user.create(session, db_obj)
    return new_user


# ----------------------------
# Read a specific user
# ----------------------------
@users_router.get("/{user_id}/", response_model=UserResponse)
async def read_user(
    user_id: int,
    current_user: User = Depends(fetch_current_user),
    session: AsyncSession = Depends(provide_session)
):
    """
    Get a specific user by id.
    """
    if current_user.id == user_id:
        return current_user

    user = await crud_user.get(session, id=user_id)
    if not current_user.is_superuser:
        raise HTTPException(
            status_code=403,
            detail="The user doesn't have enough privileges"
        )
    if not user:
        raise HTTPException(status_code=404, detail="User not found")
    return user


# ----------------------------
# Update a user (superuser only)
# ----------------------------
@users_router.put("/{user_id}/", response_model=UserResponse, dependencies=[Depends(on_superuser)])
async def update_user(
    user_id: int,
    user_in: UserUpdateSchema,
    session: AsyncSession = Depends(provide_session)
):
    user = await crud_user.get(session, id=user_id)
    if not user:
        raise HTTPException(
            status_code=404,
            detail="The user with this ID does not exist in the system"
        )
    try:
        update_data = user_in.dict(exclude_unset=True, exclude_none=True)
        if user_in.password:
            update_data["hashed_password"] = hash_password(user_in.password)
        user = await crud_user.update(session, db_obj=user, obj_in=update_data)
    except IntegrityError:
        raise HTTPException(
            status_code=409,
            detail="User with this email already exists"
        )
    return user


# ----------------------------
# Delete a user
# ----------------------------
@users_router.delete("/{user_id}/", status_code=204)
async def delete_user(
    user_id: int,
    current_user: User = Depends(on_superuser),
    session: AsyncSession = Depends(provide_session)
):
    user = await crud_user.get(session, id=user_id)
    if not user:
        raise HTTPException(status_code=404, detail="User not found")
    if current_user.id == user_id:
        raise HTTPException(status_code=403, detail="User can't delete itself")
    await crud_user.delete(session, db_obj=user)
```

```python
# ----------------------------
# ARQ Task Queue / Worker Setup
# ----------------------------
import asyncio
import os
import uvloop
from arq import create_pool
from arq.connections import RedisSettings

# Use uvloop for faster asyncio loop
asyncio.set_event_loop_policy(uvloop.EventLoopPolicy())

# Environment variables (fallback to defaults)
REDIS_HOST = os.getenv("REDIS_HOST", "localhost")
REDIS_PORT = int(os.getenv("REDIS_PORT", 6379))

# ----------------------------
# ARQ Redis Pool
# ----------------------------
from __main__ import redis_pool, settings

async def create_redis_pool() -> None:
    """Create ARQ Redis pool on app startup."""
    global redis_pool
    if redis_pool is None:
        redis_pool = await create_pool(
            RedisSettings(host=settings.REDIS_HOST, port=settings.REDIS_PORT)
        )

async def close_redis_pool() -> None:
    """Close ARQ Redis pool on app shutdown."""
    global redis_pool
    if redis_pool:
        redis_pool.close()
        await redis_pool.wait_closed()
        redis_pool = None

# ----------------------------
# Sample ARQ task
# ----------------------------
async def test_task(ctx, word: str) -> str:
    """Example background task."""
    await asyncio.sleep(10)
    return f"test task return {word}"

async def startup(ctx):
    print("ARQ worker startup...")

async def shutdown(ctx):
    print("ARQ worker shutdown...")

# ----------------------------
# ARQ Worker Settings
# ----------------------------
class WorkerSettings:
    functions = [test_task]
    redis_settings = RedisSettings(host=REDIS_HOST, port=REDIS_PORT)
    on_startup = startup
    on_shutdown = shutdown
    handle_signals = False

# ----------------------------
# FastAPI Application Factory
# ----------------------------
from fastapi import FastAPI
from __main__ import health_router, home_router, auth_router, tasks_router, users_router

def create_application() -> FastAPI:
    """Create FastAPI app and include routers with ARQ startup/shutdown handlers."""
    app = FastAPI(title=settings.PROJECT_NAME)

    # Include all routers
    app.include_router(health_router)
    app.include_router(home_router)
    app.include_router(auth_router)
    app.include_router(tasks_router)
    app.include_router(users_router)

    # Add ARQ Redis pool startup/shutdown
    app.add_event_handler("startup", create_redis_pool)
    app.add_event_handler("shutdown", close_redis_pool)

    return app

# Create global FastAPI app
app = create_application()
```

```python
import asyncio
from fastapi import FastAPI, APIRouter, HTTPException
from pydantic import BaseModel
from httpx import AsyncClient, ASGITransport

# ----------------------------
# Mock Redis with proper methods
# ----------------------------
class MockRedis:
    def __init__(self):
        self.store = []

    async def lpush(self, key, value):
        self.store.append(value)
        return len(self.store) - 1  # return index as job_id

    async def lindex(self, key, index):
        try:
            return self.store[index]
        except IndexError:
            return None

redis_pool = MockRedis()

# ----------------------------
# Task Schema
# ----------------------------
class TaskCreate(BaseModel):
    message: str

# ----------------------------
# Tasks Router
# ----------------------------
tasks_router = APIRouter(prefix="/tasks", tags=["Tasks"])

@tasks_router.post("/", response_model=dict)
async def enqueue_task(task: TaskCreate):
    job_id = await redis_pool.lpush("task_queue", task.message)
    return {"id": str(job_id)}

@tasks_router.get("/{task_id}/")
async def fetch_task_info(task_id: str):
    job_data = await redis_pool.lindex("task_queue", int(task_id))
    if job_data is None:
        raise HTTPException(status_code=404, detail="Task not found")
    return {"task_id": task_id, "message": job_data}

# ----------------------------
# FastAPI app
# ----------------------------
app = FastAPI()
app.include_router(tasks_router)

# ----------------------------
# Test function
# ----------------------------
async def test_tasks():
    transport = ASGITransport(app=app)
    async with AsyncClient(transport=transport, base_url="http://testserver") as client:

        # Enqueue first task
        response = await client.post("/tasks/", json={"message": "hello world"})
        print("Enqueue Response:", response.json())
        job_id = response.json()["id"]

        # Fetch first task
        response2 = await client.get(f"/tasks/{job_id}/")
        print("Fetch Response:", response2.json())

        # Fetch non-existent task
        response3 = await client.get("/tasks/999/")
        print("Non-existent Response Status:", response3.status_code)
        print("Non-existent Response Body:", response3.json())

        # Enqueue second task
        response4 = await client.post("/tasks/", json={"message": "second task"})
        print("Second Enqueue Response:", response4.json())

        # Fetch second task
        response5 = await client.get(f"/tasks/{response4.json()['id']}/")
        print("Second Fetch Response:", response5.json())

# ----------------------------
# Run the test
# ----------------------------
asyncio.run(test_tasks())
```
